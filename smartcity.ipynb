Import Library Python

```python
import pandas as pd
import matplotlib.pyplot as plt
```

Membaca Dataset

```python
data = pd.read_csv('smart_mobility_dataset.csv')

data.head()
```

Mengecek Struktur Dataset

```python
data.info()
```

Menghapus Data Kosong

```python
data = data.dropna()
```

Menghapus Data Duplikat

```python
data = data.drop_duplicates()
```

Melihat Statistik Dataset

```python
data.describe()
```

Analisis Jumlah Kendaraan

```python
data['Vehicle_Count'].mean()
```

Analisis Kepadatan Traffic

```python
data.sort_values(by='Road_Occupancy_%', ascending=False)
```

Analisis Kecepatan Kendaraan

```python
data['Traffic_Speed_kmh'].mean()
```

Grafik Vehicle Count

```python
data['Vehicle_Count'].plot(kind='hist')

plt.title('Distribusi Jumlah Kendaraan')

plt.xlabel('Jumlah Kendaraan')

plt.ylabel('Frekuensi')

plt.show()
```

Menyimpan Hasil Analisis

```python
data.to_csv('hasil_smartcity.csv', index=False)
```
